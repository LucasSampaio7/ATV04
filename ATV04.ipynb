```python
#Importando o módulo statistics e criando uma lista com 14 elementos em ordem crescente
import statistics

dados = [10, 12, 14, 15, 18, 20, 22, 25, 28, 30, 32, 35, 38, 40]
```

```python
#Calculando a média e a mediana do módulo
media_1 = statistics.mean(dados)
mediana_1 = statistics.median(dados)
print("Usando import statistics:")
print("Média:", media_1)
print("Mediana:", mediana_1)
```

```python
#Usando um alias (apelido) módulo e refazendo os cálculos
import statistics as st

media_2 = st.mean(dados)
mediana_2 = st.median(dados)
print("\nUsando import statistics as st:")
print("Média:", media_2)
print("Mediana:", mediana_2)
```

```python
#Importando apenas as funções mean e median
from statistics import mean, median

media_3 = mean(dados)
mediana_3 = median(dados)
print("\n4 - Usando from statistics import mean, median:")
print("Média:", media_3)
print("Mediana:", mediana_3)
```

```python
#Importanto todas as funções do módulo statistics
from statistics import *

media_4 = mean(dados)
mediana_4 = median(dados)
print("\nUsando from statistics import *:")
print("Média:", media_4)
print("Mediana:", mediana_4)
```
